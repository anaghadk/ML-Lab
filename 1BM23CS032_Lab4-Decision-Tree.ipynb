**Comparison: Regression Trees vs. Decision Tree Classifiers**

Decision Trees, whether for classification or regression, work by recursively partitioning the data space. However, their handling of the target variable and splitting criteria differ fundamentally:

1.  **Target Variable:**
    *   **Decision Tree Classifiers:** Handle *categorical* or *discrete* target variables. The goal is to assign an input to one of a finite set of classes.
    *   **Decision Tree Regressors:** Handle *continuous* or *numerical* target variables. The goal is to predict a specific numerical value.

2.  **Splitting Criteria (Impurity Measures):**
    *   **Decision Tree Classifiers:** Typically use impurity measures like **Gini Impurity** or **Entropy** to decide the best split. The objective is to maximize the homogeneity of the target variable within each child node after a split. A node is considered "pure" if all samples in it belong to the same class.
    *   **Decision Tree Regressors:** Typically use measures like **Mean Squared Error (MSE)**, **Mean Absolute Error (MAE)**, or **Friedman MSE** (an improvement over MSE for some trees). The objective is to minimize the variance or error of the target variable within each child node. A node is considered "pure" if all samples in it have similar target values.

3.  **Leaf Node Predictions:**
    *   **Decision Tree Classifiers:** The prediction for a new instance that falls into a specific leaf node is typically the **majority class** of the training samples in that leaf node (or sometimes the probability distribution over classes).
    *   **Decision Tree Regressors:** The prediction for a new instance that falls into a specific leaf node is typically the **average (mean)** or **median** of the target values of the training samples in that leaf node.
